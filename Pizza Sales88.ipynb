```python
#feature engineering
```

```python
df['order_date']=pd.to_datetime(df['order_date'],format='mixed')
df['order_time']=pd.to_datetime(df['order_time'],format='mixed')
```

```python
df['order_hour']=df['order_time'].dt.hour#extract hour
df['order_month']=df['order_date'].dt.month_name()#extract month name
df['month_id']=df['order_date'].dt.month#extract month number
df['order_day']=df['order_date'].dt.day_name()#extract day name
```

```python
day_num={
    'Sunday':1,
    'Monday':2,
    'Tuesday':3,
    'Wednesday':4,
    'Thursday':5,
    'Friday':6,
    'Saturday':7
}
df['day_id']=df['order_day'].map(day_num)
    
```

```python
#function banaune
def currencyConverter(dataframe,*args):
    for col in args:
        dataframe[col]=dataframe[col].map("Rs.{:,.2f}".format)
def pctConverter(dataframe, *args):
    for col in args:
        dataframe[col]=dataframe[col].map("Rs.{:,.2f}%".format)
        
```

```python
'''Key Performance Indicators (KPIs) 
Total Revenue = Sum of total_price 
Total Pizzas Sold = Sum of quantity 
Total Orders = Count of unique order_id 
Average Order Value (AOV) = Total Revenue ÷ Total Orders 
Average Pizza per Order = Total Pizzas Sold ÷ Total Orders '''
#KPI'S
```

```python
total_revenue=df['total_price'].sum()
total_revenue
```

```python
total_revenue=df['total_price'].sum()
total_pizza_sold=df['quantity'].sum()
#nunique->number of total unique orders
total_orders=df['order_id'].nunique()
average_order_value=total_revenue/total_orders
average_pizza_per_order=total_pizza_sold/total_orders
```

```python
print(f'Total Revenue : Rs.{total_revenue:,.2f}')
print(f'Total Pizza Sold:{total_pizza_sold}qty.')
print(f'Total Orders: {total_orders}.')
print(f'Average Order Value:Rs {average_order_value:,.2f}')
print(f'Average Pizza Per Order: {average_pizza_per_order:,.2f}.')
```

```python
#analysis
##daily trend sales by day of week
daily_trend=df.groupby(['day_id','order_day'])['total_price'].sum().reset_index().rename(
    columns={'order_day': 'Order Day','total_price':'Total Price'}
)
currencyConverter(daily_trend,'Total Price')
daily_trend[['Order Day','Total Price']]
```

```python
####hourly trend: sales by hour of day
hourly_trend=df.groupby('order_hour')['total_price'].sum().reset_index().rename(
    columns={'order_hour':'Order Hour','total_price':'Total Price'}
)
currencyConverter(hourly_trend,'Total Price')
hourly_trend[['Order Hour','Total Price']]
```

```python
#monthly trend :monthly revenue and orders
monthly_trend=df.groupby(['month_id','order_month']).agg(
    total_price=('total_price','sum'),
    total_orders=('order_id','nunique')
).reset_index().rename(
    columns={'order_month':'Order Month','total_price':'Total Price','total_orders':'Total Orders'}
)
currencyConverter(monthly_trend,'Total Price')
monthly_trend[['Order Month','Total Price','Total Orders']]
```

```python
#new day june 22
```

```python

#% of sales by category
cat_sales_pct = df.groupby('pizza_category')['total_price'].sum().reset_index()
cat_sales_pct['sales_pct'] = (cat_sales_pct['total_price'] / total_revenue)*100
currencyConverter(cat_sales_pct, 'total_price')
pctConverter(cat_sales_pct, 'sales_pct')
cat_sales_pct.rename(columns = {
    'pizza_category': 'Pizza Category',
    'total_price': 'Total Price',
    'sales_pct': 'Sales Pct'
})
```

```python
#% of sales by pizza size and category
size_cat_sales_pct = df.groupby(['pizza_size', 'pizza_category'])['total_price'].sum().reset_index()
size_cat_sales_pct['sales_pct'] = (size_cat_sales_pct['total_price'] / total_revenue)*100

currencyConverter(size_cat_sales_pct, 'total_price')
pctConverter(size_cat_sales_pct, 'sales_pct')
size_cat_sales_pct.rename(columns = {
    'pizza_size': 'Pizza Size',
    'pizza_category': 'Pizza Category',
    'total_price': 'Total Price',
    'sales_pct': 'Sales Pct'
})
```

```python
#total pizza sold by pizza category
pizza_sold_cat = df.groupby('pizza_category')['quantity'].sum().reset_index().rename(
    columns = {'pizza_category': 'Pizza Category', 'quantity': 'Total Pizza Sold'}
)
pizza_sold_cat['Total Pizza Sold'] = pizza_sold_cat['Total Pizza Sold'].map("{:,.2f} qty".format)
pizza_sold_cat
```

```python
# top-5 best selling pizzas 
top_5=df.groupby('pizza_name').agg(
    total_revenue=('total_price','sum'),
    total_orders=('order_id','nunique'),
    total_sold=('quantity','sum'),
).reset_index().sort_values(ascending=False,by='total_revenue',ignore_index=True).head()
currencyConverter(top_5,'total_revenue')
top_5['total_orders']=top_5['total_orders'].map('{:,.0f}'.format)
top_5['total_sold']=top_5['total_sold'].map('{:,.0f} qty'.format)
top_5.rename(columns={
    'pizza_name':'Pizza Name','total_revenue':'Total Revenue','total_orders':'Total Orders',
    'total_sold':'Total Sold'
})
```

```python
#bottom 5 least selling pizzas
bottom_5 = df.groupby('pizza_name').agg(
    total_revenue = ('total_price', 'sum')
).reset_index().sort_values(ascending=True, by='total_revenue', ignore_index=True).head()

currencyConverter(bottom_5, 'total_revenue')
bottom_5.rename(columns = {
    'pizza_name': 'Pizza Name', 'total_revenue': 'Total Revenue'
})
```
